## Data and method

Generate four random images with two synthetic captions each. Split by image ID into
training, validation, and test partitions. Fit the vocabulary on training captions only.
The real encoder architecture uses random weights; reduced recurrent widths keep this
CPU smoke test practical. Real training uses the original 256/300 widths.
The pipeline saves vocabulary and preprocessing configuration alongside the model.


```python
from pathlib import Path
from captioning.pipeline import run
encoder = 'mobilenet'
reports = {variant: run(output=f"artifacts/notebook-{encoder}-{variant}", encoder=encoder, variant=variant)
           for variant in ("lstm", "blstm")}
for variant, report in reports.items():
    print(variant, report["status"])
    print("Feature dimension:", report["configuration"]["feature_dim"])
    print("Training loss:", report["history"][0]["train"]["loss"])
    print("Generated caption:", report["predictions"])
```

## Evaluation

Evaluation receives each model explicitly. Mean sentence BLEU-1 approximates the original
aggregation; corpus BLEU-1 through BLEU-4 are reported separately. All use unsmoothed NLTK
BLEU, whitespace tokens, and a 0–1 scale. Scores on these random fixtures have no performance
meaning; the original 0.57–0.59 claims are historical and unverified.
